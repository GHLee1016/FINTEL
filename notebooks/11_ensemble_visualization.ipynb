## 1. Load results and check headline numbers

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve()

ensemble_results = pd.read_csv(PROJECT_ROOT / 'results' / 'ensemble_results_equal_weight.csv')
comparison = pd.read_csv(PROJECT_ROOT / 'results' / 'ensemble_best_comparison.csv')
prediction_log = pd.read_csv(PROJECT_ROOT / 'log' / 'Ensemble' / 'ensemble_prediction_log.csv')

comparison['ensemble_better_RMSE_CV'] = comparison['RMSE_CV_improvement'] > 0

print(f'ensemble_results: {len(ensemble_results):,} rows')
print(f'best_comparison: {len(comparison):,} rows')
print(f'ensemble_prediction_log: {len(prediction_log):,} rows')
print()
print(f"QLIKE improved: {int(comparison['ensemble_better_QLIKE'].sum())}/{len(comparison)}")
print(f"RMSE_CV improved: {int(comparison['ensemble_better_RMSE_CV'].sum())}/{len(comparison)}")

display(comparison.head())
```
